```python
import serial
import time
```

```python
READ_DELAY = 0.05
SERIAL_TIMEOUT = 1.0
```

```python
s = serial.Serial(
    port="COM15",
    baudrate=19200,
    bytesize=8,
    parity="N",
    stopbits=1,
    timeout=SERIAL_TIMEOUT,
)
```

```python
s.write(b"A\r")
lines = []
time.sleep(READ_DELAY)
lines += s.readlines()
print(lines)
```

```python
s.read_all()
```
